```python
from langchain_community.document_loaders import UnstructuredPDFLoader
from langchain_core.documents import Document

# 1. Load the PDF using Unstructured in "elements" mode
# This breaks the PDF into detailed distinct parts (Titles, Text, List Items)
loader = UnstructuredPDFLoader(
    "./Machine_learning_Wikipedia.pdf",
    mode="elements",
    strategy="fast", # Use "hi_res" if the PDF has complex layout/columns, but "fast" is better for text-heavy Wikipedia
)

docs = loader.load()

print(f"Total raw elements found: {len(docs)}")

# ---------------------------------------------------------
# 2. Logic to Clean (Remove References) & Group by Header
# ---------------------------------------------------------

grouped_documents = []
current_heading = "Introduction" # Default heading for text before the first title
current_text_buffer = []

for doc in docs:
    element_type = doc.metadata.get("category", "")
    text = doc.page_content.strip()
    
    # STOP Condition: If we hit the References section, stop processing entirely
    if element_type == "Title" and "reference" in text.lower():
        print("Stopping processing at 'References' section.")
        break
    
    # NEW SECTION Condition: If we hit a Title, save previous buffer and start new
    if element_type == "Title":
        # If we have gathered text for the previous section, save it
        if current_text_buffer:
            combined_content = "\n".join(current_text_buffer)
            # Create a clean LangChain Document
            new_doc = Document(
                page_content=combined_content,
                metadata={"source": "wikipedia", "section": current_heading}
            )
            grouped_documents.append(new_doc)
        
        # Update current heading and reset buffer
        current_heading = text
        current_text_buffer = []
        
    # CONTENT Condition: If it's text or a list item, add to current buffer
    elif element_type in ["NarrativeText", "ListItem", "UncategorizedText"]:
        if len(text) > 5: # Filter out tiny artifacts/page numbers
            current_text_buffer.append(text)

# Don't forget to add the very last section after the loop finishes
if current_text_buffer:
    combined_content = "\n".join(current_text_buffer)
    new_doc = Document(
        page_content=combined_content,
        metadata={"source": "wikipedia", "section": current_heading}
    )
    grouped_documents.append(new_doc)

# ---------------------------------------------------------
# 3. Validation
# ---------------------------------------------------------

print(f"\nFinal Number of Chunks (by Section): {len(grouped_documents)}")

# Preview a few chunks to show the panel
for i, doc in enumerate(grouped_documents[:3]):
    print(f"--- Chunk {i+1} [Section: {doc.metadata['section']}] ---")
    print(doc.page_content[:200] + "...") # Print first 200 chars
    print("\n")
```
